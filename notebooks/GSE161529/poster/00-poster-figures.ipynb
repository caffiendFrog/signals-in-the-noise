# Poster figures

Two figures, in the ISCB palette from `12-a-iscb-scs-volcano.ipynb`:
Michigan navy `#00274C` for text and axes, blue `#0072B2`, bluish green
`#009E73` for ER+, vermillion `#D55E00`, and yellow-orange `#E69F00`.
Backgrounds are transparent. Each file is written as PNG and PDF at 300 dpi.

The depth figure reads notebook 04's `04_summary_by_version.csv` and
`04_depth_and_thinning.csv`. The progenitor figure reads notebook 03's
`03_curve.csv`, `03_baseline.csv`, `03_donor_shares.csv`, and `donors.csv`.

```python
import pandas as pd
from IPython.display import Image, display

from signals_in_the_noise.analysis.poster_figures import plot_depth_collapse, plot_lp_expansion
from signals_in_the_noise.analysis.qc_filtering import output_directory
from signals_in_the_noise.config import get_data_path

FILTER_DIR = output_directory()
CONFOUND_DIR = get_data_path("processed") / "qc_confound"
POSTER_DIR = get_data_path("processed") / "poster"
POSTER_DIR.mkdir(parents=True, exist_ok=True)
POSTER_DIR
```

## 1. Library depth and the ER+ signal

```python
summary = pd.read_csv(CONFOUND_DIR / "04_summary_by_version.csv")
depths = pd.read_csv(CONFOUND_DIR / "04_depth_and_thinning.csv")
depth_path = POSTER_DIR / "poster-depth-collapse.png"
plot_depth_collapse(summary, depths, depth_path)
display(Image(filename=str(depth_path)))
```

## 2. BRCA1 luminal progenitors and the removed cells

```python
curve = pd.read_csv(FILTER_DIR / "03_curve.csv")
baseline = pd.read_csv(FILTER_DIR / "03_baseline.csv").iloc[0]
shares = pd.read_csv(FILTER_DIR / "03_donor_shares.csv")
donors = pd.read_csv(FILTER_DIR / "donors.csv")
shares = shares.merge(
    donors[["donor", "menopause_status"]],
    on="donor",
    how="left",
)
lp_path = POSTER_DIR / "poster-lp-expansion.png"
plot_lp_expansion(
    curve,
    shares,
    lp_path,
    alpha=0.01,
    signature_c=float(baseline["signature_c"]),
    extreme_p=float(baseline["extreme_p"]),
)
display(Image(filename=str(lp_path)))
```
